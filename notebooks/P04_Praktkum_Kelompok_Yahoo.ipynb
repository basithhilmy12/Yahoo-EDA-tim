## Bagian 3 — Kelengkapan: nilai kosong dan penyamarannya

Ada dua macam nilai kosong, dan hanya satu yang terlihat oleh `isna()`.

| Jenis | Contoh | Terdeteksi `isna()`? |
|---|---|---|
| Kosong sejati | `NaN`, sel kosong di CSV | Ya |
| Placeholder angka | `-999`, `0` yang berarti tidak diisi, `9999` | **Tidak** |
| Placeholder teks | `N/A`, `-`, `null`, `tidak diketahui`, `?` | **Tidak** |

Placeholder berbahaya justru karena ia terlihat seperti data sungguhan.

````python
# CONTOH 3.1 — kosong sejati
kosong = df.isna().sum()
kosong[kosong > 0].to_frame('jumlah_kosong').assign(
    persen=lambda d: (d.jumlah_kosong / len(df) * 100).round(1))
````

````python
# CONTOH 3.2 — berburu placeholder teks
CURIGA = ['n/a', 'na', '-', '', 'null', 'none', 'tidak diketahui', 'unknown', '?', '#n/a']

for kol in df.select_dtypes(include='object').columns:
    nilai = df[kol].astype(str).str.strip().str.lower()
    ketemu = nilai[nilai.isin(CURIGA)].value_counts()
    if len(ketemu):
        print(f'{kol}:')
        print(ketemu.to_string(), '\n')
````

````python
# CONTOH 3.3 — berburu placeholder angka
# Caranya: lihat nilai ekstrem pada tiap kolom numerik
for kol in df.select_dtypes(include='number').columns:
    s = df[kol]
    print(f'{kol:18s} min={s.min():12,.1f}  max={s.max():14,.1f}  '
          f'nilai terkecil unik: {sorted(s.dropna().unique())[:3]}')
````

> **Pola yang perlu dihafal.** Placeholder angka hampir selalu berupa bilangan yang mencolok:
> `-1`, `-999`, `0`, `9999`, `99999`, atau tahun `1900`. Kalau `min` atau `max` sebuah kolom terasa aneh,
> periksa nilai uniknya sebelum menganggapnya data sungguhan.

## Bagian 4 — Mengapa data itu kosong: MCAR, MAR, MNAR

Menemukan nilai kosong baru separuh pekerjaan. Separuh lainnya adalah menebak **mengapa** kosong,
karena alasannya menentukan apakah Anda boleh menghapus baris, boleh mengisi, atau tidak boleh keduanya.

| Istilah | Artinya | Contoh | Akibatnya |
|---|---|---|---|
| **MCAR** | Kosong betul-betul acak | Kertas struk sobek | Menghapus baris aman, hanya kehilangan ukuran sampel |
| **MAR** | Kosong bergantung pada kolom lain **yang terlihat** | Berat barang kosong untuk produk non-fisik | Bisa ditangani; isi per kelompok, atau akui bahwa kolom itu memang tidak berlaku |
| **MNAR** | Kosong bergantung pada **nilainya sendiri** yang tidak terlihat | Saldo besar sengaja tidak ditampilkan | Paling berbahaya; mengisi rata-rata akan menyesatkan |

> **Hal terpenting di bagian ini:** MAR dapat dibuktikan dari data. MCAR dan MNAR **tidak bisa dibedakan**
> hanya dengan melihat data, karena nilai yang hilang tidak dapat diperiksa. Membedakannya memerlukan
> pengetahuan tentang bagaimana data itu dikumpulkan.

````python
# CONTOH 4.1 — alat bantu: apakah kekosongan terkait kolom lain?
def pola_kosong(data, kolom_kosong, kolom_pembanding):
    """Persentase kekosongan kolom_kosong untuk tiap nilai kolom_pembanding."""
    return (data.groupby(kolom_pembanding)[kolom_kosong]
                .apply(lambda s: round(s.isna().mean() * 100, 1))
                .sort_values(ascending=False)
                .to_frame('persen_kosong'))

# rapikan kategori lebih dulu supaya pengelompokannya bermakna
kat_rapi = (df['kategori'].astype(str).str.strip().str.title()
              .replace({'Minumam': 'Minuman', 'Alat Tulis': 'Atk'}))
df_bantu = df.assign(kategori_rapi=kat_rapi)

pola_kosong(df_bantu, 'berat_kg', 'kategori_rapi')
````

> **Ini MAR, dan buktinya jelas.** `berat_kg` kosong 100% untuk Jasa dan Pulsa — keduanya memang tidak
> punya berat. Kekosongannya bergantung pada `kategori`, kolom yang bisa kita lihat.
>
> Tindakan yang tepat **bukan** mengisi dengan rata-rata, melainkan mengakui bahwa kolom ini hanya
> berlaku untuk barang fisik. Mengisi berat Pulsa dengan 0,3 kg akan menciptakan data palsu.

````python
# CONTOH 4.2 — bandingkan dengan kolom lain
pola_kosong(df_bantu, 'catatan_kasir', 'kategori_rapi')
````

> **Ini tampak MCAR.** Persentasenya mirip di semua kategori. Tidak ada kolom teramati yang menjelaskan
> kekosongannya, jadi menghapus atau mengabaikan kolom ini relatif aman.
>
> Catatan: sebagian kekosongan ini sebenarnya berasal dari sel yang memang dibiarkan kosong oleh kasir
> (teks kosong), yang oleh `read_csv` dibaca menjadi `NaN`. Teks kosong dan data hilang adalah dua hal
> berbeda, tetapi CSV tidak bisa membedakannya. Ini keterbatasan format, bukan kesalahan Anda.

````python
# CONTOH 4.3 — kasus yang menipu
for kol in ['kategori_rapi', 'kasir', 'metode_bayar']:
    print(f'--- saldo_simpanan kosong menurut {kol} ---')
    print(pola_kosong(df_bantu, 'saldo_simpanan', kol).head(5).to_string(), '\n')
````

````python
# CONTOH 4.4 — dan pembandingan rata-rata kolom numerik
m = df['saldo_simpanan'].isna()
banding = pd.DataFrame({
    'saat_kosong': df.loc[m, ['jumlah', 'harga_satuan', 'diskon_persen']].mean().round(2),
    'saat_terisi': df.loc[~m, ['jumlah', 'harga_satuan', 'diskon_persen']].mean().round(2),
})
print(banding.to_string())
print()
print('Saldo yang TERISI — median:', f"{df['saldo_simpanan'].median():,.0f}",
      '| maksimum:', f"{df['saldo_simpanan'].max():,.0f}")
````

> **Baca pelan-pelan, ini inti pertemuan hari ini.**
>
> Setiap pemeriksaan di atas mengatakan kekosongan `saldo_simpanan` tidak terkait kolom mana pun.
> Secara statistik, kolom ini **tampak MCAR**, dan Anda akan tergoda mengisinya dengan median.
>
> Sekarang perhatikan satu hal yang tidak muncul di pemeriksaan mana pun: saldo tertinggi yang
> *masih terlihat* hanya sekitar 3,5 juta. Untuk koperasi dengan ratusan anggota, tidak adanya satu pun
> saldo besar adalah hal yang ganjil. Dugaan yang masuk akal: anggota bersaldo besar memilih tidak
> menampilkan saldonya.
>
> Kalau dugaan itu benar, kolom ini **MNAR**, dan mengisinya dengan median akan membuat rata-rata saldo
> koperasi tampak jauh lebih kecil daripada kenyataannya. Tidak ada satu pun uji statistik yang bisa
> memastikan hal ini. Yang bisa memastikannya hanya satu: **bertanya kepada pemilik data**.

## Bagian 5 — Keunikan: duplikat persis dan duplikat nyaris

Duplikat persis mudah ditemukan. Yang merepotkan adalah duplikat nyaris: baris yang merujuk kejadian
yang sama tetapi berbeda sedikit, misalnya beda huruf besar-kecil atau satu kolom kosong.

````python
# CONTOH 5.1 — duplikat persis
print('Baris yang seluruh kolomnya sama:', df.duplicated().sum())
df[df.duplicated(keep=False)].sort_values('id_transaksi').head(6)
````

````python
# CONTOH 5.2 — duplikat pada kunci utama
# id_transaksi seharusnya unik. Kalau tidak, ada yang salah.
dup_id = df['id_transaksi'].duplicated(keep=False)
print('Baris dengan id_transaksi berulang:', dup_id.sum())
print('id_transaksi kosong             :', df['id_transaksi'].isna().sum())

contoh = df[dup_id].sort_values('id_transaksi').head(4)
contoh[['id_transaksi', 'kasir', 'catatan_kasir', 'total_bayar']]
````

````python
# CONTOH 6.3 — spasi tersembunyi pada kolom kunci
a = df['id_anggota'].astype(str)
print('id_anggota unik apa adanya :', a.nunique())
print('id_anggota unik setelah strip:', a.str.strip().nunique())
print()
print('Selisih', a.nunique() - a.str.strip().nunique(),
      'anggota hantu — muncul hanya karena ada spasi di akhir teks.')
````

> **Dampaknya nyata.** Kalau Anda menghitung jumlah anggota aktif koperasi tanpa membuang spasi,
> laporan Anda akan menyebut puluhan anggota yang sebenarnya tidak ada. Angkanya rapi, tabelnya cantik,

### Bagian paling berbahaya di seluruh workbook ini

Cara yang paling sering dipakai orang untuk menangani format campur adalah `format='mixed'`
ditambah `dayfirst=True`, karena di Indonesia tanggal memang ditulis hari lebih dulu.
Jalankan sel berikut dan amati baik-baik.

````python
# CONTOH 7.3 — cara yang tampak benar
cara_umum = pd.to_datetime(t, format='mixed', dayfirst=True, errors='coerce')

print('Gagal dikonversi :', int(cara_umum.isna().sum()))
print('Rentang tanggal  :', cara_umum.min(), 'sampai', cara_umum.max())
print('Bulan yang muncul:', sorted(cara_umum.dt.month.dropna().unique()))
````

> **Nol galat. Tidak ada peringatan apa pun. Dan hasilnya salah.**
>
> Dataset ini berisi transaksi **Agustus sampai November 2025** — empat bulan. Hasil di atas
> menyebar ke seluruh dua belas bulan.
>
> Penyebabnya: `dayfirst=True` ikut diterapkan pada baris yang formatnya sudah `yyyy-mm-dd`.
> Baris `2025-08-01` dibaca sebagai tahun 2025, **hari** 08, **bulan** 01, sehingga menjadi 8 Januari.
> Karena 08 dan 01 keduanya angka yang sah untuk hari maupun bulan, pandas tidak punya alasan mengeluh.
>
> Inilah jenis kesalahan yang paling mahal dalam analisis data: **salah yang tidak menimbulkan galat.**
> Tidak ada yang berwarna merah, tidak ada yang berhenti, dan laporan Anda tetap terbit.

````python
# CONTOH 7.4 — cara yang benar: tangani tiap format secara terpisah
benar = pd.Series(pd.NaT, index=t.index, dtype='datetime64[ns]')
benar[iso]  = pd.to_datetime(t[iso],  format='%Y-%m-%d %H:%M:%S', errors='coerce')
benar[~iso] = pd.to_datetime(t[~iso], format='%d/%m/%Y %H:%M',    errors='coerce')

print('Gagal dikonversi :', int(benar.isna().sum()))
print('Rentang tanggal  :', benar.min(), 'sampai', benar.max())
print('Bulan yang muncul:', sorted(benar.dt.month.dropna().unique()))
print()
print('Baris yang berbeda antara cara umum dan cara benar:', int((cara_umum != benar).sum()))
````

> **Pelajaran yang dibawa pulang:** setelah setiap konversi tanggal, jalankan satu uji akal sehat.
> Yang paling murah adalah memeriksa bulan atau tahun yang muncul:
>
> ```python
> print(sorted(kolom.dt.month.unique()))
> print(kolom.min(), kolom.max())
> ```
>
> Kalau rentangnya tidak sesuai dengan yang Anda ketahui tentang data itu, konversinya salah —
> berapa pun jumlah galat yang muncul.

````python
# 10.1 — Muat dataset proyek tim

import pandas as pd
import numpy as np
import os

df_tim = pd.read_csv('transaksi_koperasi_2025.csv')

print("BERHASIL MEMUAT DATASET")
print("=" * 60)
print("Ukuran dataset:", df_tim.shape)
print("\n5 baris pertama:")
display(df_tim.head())

print("\nInformasi dataset:")
df_tim.info()

print("\nRingkasan statistik numerik:")
display(df_tim.describe().round(2))

print("\nRingkasan kolom teks:")
display(df_tim.describe(include='object').T)
````

````python
# 10.2 — Sembilan pemeriksaan kualitas data

# Menyiapkan tempat untuk mencatat seluruh temuan
temuan_tim = []

def catat_tim(dimensi, kolom, masalah, jumlah, keparahan, rencana):
    temuan_tim.append({
        'dimensi': dimensi,
        'kolom': kolom,
        'masalah': masalah,
        'jumlah': int(jumlah),
        'persen': round(int(jumlah) / len(df_tim) * 100, 2),
        'keparahan': keparahan,
        'rencana': rencana
    })

# =========================================================
# 1. POTRET AWAL DATA
# =========================================================
print("=" * 65)
print("1. POTRET AWAL DATA")
print("=" * 65)
print("Jumlah baris dan kolom:", df_tim.shape)
print("\nTipe data:")
print(df_tim.dtypes)
print("\nStatistik numerik:")
display(df_tim.describe().round(2))

# =========================================================
# 2. NILAI KOSONG SEJATI
# =========================================================
print("\n" + "=" * 65)
print("2. NILAI KOSONG SEJATI")
print("=" * 65)

kosong_tim = df_tim.isna().sum()
persen_kosong_tim = (kosong_tim / len(df_tim) * 100).round(2)

ringkasan_kosong = pd.DataFrame({
    'jumlah_kosong': kosong_tim,
    'persen_kosong': persen_kosong_tim
})

display(ringkasan_kosong[ringkasan_kosong['jumlah_kosong'] > 0]
        .sort_values('persen_kosong', ascending=False))

# =========================================================
# 3. PLACEHOLDER TEKS DAN ANGKA
# =========================================================
print("\n" + "=" * 65)
print("3. PLACEHOLDER TEKS DAN ANGKA")
print("=" * 65)

placeholder_teks = [
    'n/a', 'na', '-', '', 'null', 'none',
    'tidak diketahui', 'unknown', '?', '#n/a'
]

print("Placeholder teks:")
for kol in df_tim.select_dtypes(include='object').columns:
    nilai = df_tim[kol].astype('string').str.strip().str.lower()
    jumlah_placeholder = int(nilai.isin(placeholder_teks).sum())

    if jumlah_placeholder > 0:
        print(f"{kol}: {jumlah_placeholder} baris")
        catat_tim(
            'Kelengkapan', kol,
            'Placeholder teks yang menyamarkan nilai kosong',
            jumlah_placeholder, 'Sedang',
            'Verifikasi placeholder dan tandai sebagai nilai kosong jika sesuai'
        )

print("\nPemeriksaan nilai angka yang mencurigakan:")
for kol in df_tim.select_dtypes(include='number').columns:
    s = df_tim[kol]
    print(f"{kol}: min={s.min()}, max={s.max()}")

    # -999 merupakan kode khusus yang diketahui pada rating_layanan
    if kol == 'rating_layanan':
        jumlah_sentinel = int((s == -999).sum())
        print("Jumlah rating -999:", jumlah_sentinel)

        if jumlah_sentinel > 0:
            catat_tim(
                'Kelengkapan', kol,
                'Placeholder angka -999',
                jumlah_sentinel, 'Tinggi',
                'Verifikasi lalu perlakukan sebagai nilai rating yang tidak tersedia'
            )

# =========================================================
# 4. POLA MISSING VALUE: MCAR, MAR, ATAU MNAR
# =========================================================
print("\n" + "=" * 65)
print("4. POLA MISSING VALUE")
print("=" * 65)

# Bandingkan missingness dengan beberapa kolom yang teramati.
for kol in df_tim.columns:
    persen = df_tim[kol].isna().mean() * 100

    if persen > 5:
        print(f"\nKolom: {kol} ({persen:.2f}% kosong)")

        for pembanding in ['kategori', 'metode_bayar', 'kasir']:
            if pembanding != kol:
                pola = (
                    df_tim.groupby(pembanding, dropna=False)[kol]
                    .apply(lambda s: round(s.isna().mean() * 100, 1))
                    .sort_values(ascending=False)
                )
                print(f"Persentase kosong menurut {pembanding}:")
                display(pola.to_frame('persen_kosong'))

# Interpretasi awal khusus dataset transaksi koperasi:
print("""
INTERPRETASI AWAL (HIPOTESIS, BUKAN BUKTI FINAL):
- berat_kg: MAR. Berat tidak berlaku untuk kategori Jasa dan Pulsa.
  Kekosongan berkaitan dengan kategori transaksi.
- catatan_kasir: tampak seperti MCAR jika tidak ada pola yang jelas
  pada kolom pembanding. Mekanisme ini belum bisa dipastikan.
- saldo_simpanan: tampak seperti MCAR jika tidak berkaitan dengan
  kolom teramati. Namun MNAR tetap mungkin jika saldo yang tidak
  tercatat berkaitan dengan kondisi saldo itu sendiri.

Klasifikasi harus ditinjau kembali berdasarkan hasil perbandingan
dan pengetahuan proses pencatatan koperasi.
""")

# =========================================================
# 5. DUPLIKAT PERSIS DAN DUPLIKAT KUNCI
# =========================================================
print("\n" + "=" * 65)
print("5. DUPLIKAT")
print("=" * 65)

duplikat_persis = int(df_tim.duplicated().sum())
print("Baris duplikat persis:", duplikat_persis)

if duplikat_persis > 0:
    catat_tim(
        'Keunikan', 'seluruh kolom', 'Baris duplikat persis',
        duplikat_persis, 'Sedang',
        'Telusuri apakah pencatatan ganda; jangan hapus sebelum diverifikasi'
    )

if 'id_transaksi' in df_tim.columns:
    duplikat_id = int(
        df_tim['id_transaksi'].duplicated(keep=False).sum()
    )
    print("Baris dengan ID transaksi berulang:", duplikat_id)

    if duplikat_id > 0:
        catat_tim(
            'Keunikan', 'id_transaksi',
            'ID transaksi berulang',
            duplikat_id, 'Tinggi',
            'Periksa transaksi yang memakai ID sama dan validasi sumbernya'
        )

# =========================================================
# 6. KONSISTENSI KOLOM TEKS
# =========================================================
print("\n" + "=" * 65)
print("6. KONSISTENSI KOLOM TEKS")
print("=" * 65)

for kol in df_tim.select_dtypes(include='object').columns:
    nilai_asli = df_tim[kol].nunique(dropna=True)
    nilai_rapi = (
        df_tim[kol].astype('string').str.strip().str.lower()
        .nunique(dropna=True)
    )

    print(
        f"{kol}: unik asli={nilai_asli}, "
        f"unik setelah strip + lowercase={nilai_rapi}"
    )

# Pemeriksaan khusus kategori
if 'kategori' in df_tim.columns:
    kategori_asli = df_tim['kategori'].nunique(dropna=True)

    kategori_rapi = (
        df_tim['kategori'].astype('string')
        .str.strip().str.title()
        .replace({'Minumam': 'Minuman', 'Alat Tulis': 'Atk'})
    )

    kategori_bersih = kategori_rapi.nunique(dropna=True)
    print("\nKategori unik asli:", kategori_asli)
    print("Kategori unik setelah normalisasi:", kategori_bersih)

    if kategori_asli != kategori_bersih:
        catat_tim(
            'Konsistensi', 'kategori',
            'Perbedaan kapitalisasi, spasi, atau penamaan kategori',
            kategori_asli - kategori_bersih, 'Tinggi',
            'Standarkan penulisan kategori setelah memverifikasi sinonim'
        )

# =========================================================
# 7. TIPE DATA YANG SALAH
# =========================================================
print("\n" + "=" * 65)
print("7. TIPE DATA")
print("=" * 65)

print("Tipe tanggal_transaksi:", df_tim['tanggal_transaksi'].dtype)
print("Tipe total_bayar:", df_tim['total_bayar'].dtype)

# Deteksi format tanggal secara terpisah agar tidak salah tafsir
tanggal = df_tim['tanggal_transaksi'].astype('string').str.strip()
pola_iso = tanggal.str.match(r'^\d{4}-', na=False)

tanggal_iso = pd.to_datetime(
    tanggal[pola_iso],
    format='%Y-%m-%d %H:%M:%S',
    errors='coerce'
)

tanggal_dmy = pd.to_datetime(
    tanggal[~pola_iso],
    format='%d/%m/%Y %H:%M',
    errors='coerce'
)

tanggal_parsed = pd.Series(
    pd.NaT, index=df_tim.index, dtype='datetime64[ns]'
)
tanggal_parsed.loc[tanggal_iso.index] = tanggal_iso
tanggal_parsed.loc[tanggal_dmy.index] = tanggal_dmy

gagal_tanggal = int(tanggal_parsed.isna().sum())
print("Tanggal gagal dikonversi dengan format terpisah:", gagal_tanggal)

jumlah_rp = int(
    df_tim['total_bayar'].astype('string')
    .str.startswith('Rp', na=False).sum()
)
print("Total bayar berformat teks Rp:", jumlah_rp)

if gagal_tanggal > 0:
    catat_tim(
        'Validitas', 'tanggal_transaksi',
        'Tanggal gagal dikenali sesuai format yang diharapkan',
        gagal_tanggal, 'Tinggi',
        'Periksa format tanggal yang belum teridentifikasi'
    )

if jumlah_rp > 0:
    catat_tim(
        'Konsistensi', 'total_bayar',
        'Nilai uang bercampur antara angka dan format teks Rp',
        jumlah_rp, 'Tinggi',
        'Seragamkan representasi mata uang pada tahap pembersihan'
    )

# =========================================================
# 8. ATURAN RENTANG: MINIMAL LIMA ATURAN VALIDITAS
# =========================================================
print("\n" + "=" * 65)
print("8. ATURAN VALIDITAS")
print("=" * 65)

aturan_tim = {
    'Jumlah harus positif': df_tim['jumlah'] <= 0,
    'Harga satuan harus positif': df_tim['harga_satuan'] <= 0,
    'Diskon harus 0 sampai 100': (
        (df_tim['diskon_persen'] < 0) |
        (df_tim['diskon_persen'] > 100)
    ),
    'Rating harus 1 sampai 5': (
        (df_tim['rating_layanan'] < 1) |
        (df_tim['rating_layanan'] > 5)
    ),
    'Berat tidak boleh negatif': df_tim['berat_kg'] < 0,
    'Saldo simpanan tidak boleh negatif': df_tim['saldo_simpanan'] < 0,
    'ID transaksi tidak boleh kosong': (
        df_tim['id_transaksi'].isna() |
        df_tim['id_transaksi'].astype('string').str.strip().eq('')
    )
}

for nama, kondisi in aturan_tim.items():
    jumlah_pelanggar = int(kondisi.fillna(False).sum())
    print(f"{nama}: {jumlah_pelanggar} baris melanggar")

    if jumlah_pelanggar > 0:
        kolom = {
            'Jumlah harus positif': 'jumlah',
            'Harga satuan harus positif': 'harga_satuan',
            'Diskon harus 0 sampai 100': 'diskon_persen',
            'Rating harus 1 sampai 5': 'rating_layanan',
            'Berat tidak boleh negatif': 'berat_kg',
            'Saldo simpanan tidak boleh negatif': 'saldo_simpanan',
            'ID transaksi tidak boleh kosong': 'id_transaksi'
        }[nama]

        catat_tim(
            'Validitas', kolom, nama,
            jumlah_pelanggar, 'Tinggi',
            'Verifikasi aturan bisnis dan telusuri baris yang melanggar'
        )

# =========================================================
# 9. ATURAN HUBUNGAN ANTARKOLOM
# =========================================================
print("\n" + "=" * 65)
print("9. ATURAN HUBUNGAN ANTARKOLOM")
print("=" * 65)

# Rumus bisnis:
# total_bayar = jumlah * harga_satuan * (1 - diskon_persen / 100)
# Untuk mengikuti latihan, pemeriksaan utama dilakukan pada nilai
# total_bayar yang tidak diawali "Rp".

mask_angka = ~df_tim['total_bayar'].astype('string').str.startswith(
    'Rp', na=False
)

nominal = pd.to_numeric(
    df_tim.loc[mask_angka, 'total_bayar'],
    errors='coerce'
)

seharusnya = (
    df_tim.loc[mask_angka, 'jumlah'] *
    df_tim.loc[mask_angka, 'harga_satuan'] *
    (1 - df_tim.loc[mask_angka, 'diskon_persen'] / 100)
).round()

bisa_diperiksa = nominal.notna() & seharusnya.notna()
tidak_cocok = bisa_diperiksa & nominal.ne(seharusnya)

print("Baris yang bisa diperiksa:", int(bisa_diperiksa.sum()))
print("Total bayar tidak cocok:", int(tidak_cocok.sum()))

if tidak_cocok.sum() > 0:
    catat_tim(
        'Akurasi', 'total_bayar',
        'Total bayar tidak sesuai rumus jumlah, harga, dan diskon',
        int(tidak_cocok.sum()), 'Tinggi',
        'Telusuri transaksi dan verifikasi aturan perhitungan total'
    )

display(
    df_tim.loc[mask_angka].loc[
        tidak_cocok[tidak_cocok].index,
        ['id_transaksi', 'jumlah', 'harga_satuan',
         'diskon_persen', 'total_bayar']
    ].head(10)
)

print("\nSELESAI: sembilan pemeriksaan kualitas data telah dijalankan.")
print("Jumlah temuan sementara:", len(temuan_tim))
````

````python

# 10.3 — Susun profil kualitas data tim

import os
import pandas as pd

# Salin temuan yang sudah diperoleh pada sel 10.2
profil_tim = pd.DataFrame(temuan_tim).copy()

# Persentase missing value
persen_kosong = df_tim.isna().mean() * 100

# Diagnosis awal berdasarkan hasil pemeriksaan
diagnosis_missing = {
    'berat_kg': {
        'masalah': 'Missing value berkaitan dengan kategori transaksi',
        'keparahan': 'Sedang',
        'rencana': (
            'Verifikasi apakah berat memang tidak berlaku untuk '
            'transaksi jasa dan pulsa'
        )
    },
    'catatan_kasir': {
        'masalah': 'Missing value menunjukkan pola antarkategori',
        'keparahan': 'Sedang',
        'rencana': (
            'Periksa apakah catatan hanya diisi pada transaksi tertentu '
            'dan konfirmasi prosedur pencatatan'
        )
    },
    'saldo_simpanan': {
        'masalah': 'Mekanisme missing value belum dapat dipastikan',
        'keparahan': 'Sedang',
        'rencana': (
            'Bandingkan missingness berdasarkan kategori, metode bayar, '
            'dan kasir; konfirmasi penyebab kepada pengelola'
        )
    }
}

# Tambahkan kolom dengan missing value > 5%
temuan_missing = []

for kolom, persen in persen_kosong.items():
    if persen > 5:
        jumlah = int(df_tim[kolom].isna().sum())

        detail = diagnosis_missing.get(kolom, {
            'masalah': 'Missing value lebih dari 5%',
            'keparahan': 'Sedang',
            'rencana': 'Telusuri penyebab nilai kosong sebelum menentukan tindakan'
        })

        temuan_missing.append({
            'dimensi': 'Kelengkapan',
            'kolom': kolom,
            'masalah': detail['masalah'],
            'jumlah': jumlah,
            'persen': round(persen, 2),
            'keparahan': detail['keparahan'],
            'rencana': detail['rencana']
        })

# Gabungkan temuan awal dan missing value
if temuan_missing:
    profil_tim = pd.concat(
        [profil_tim, pd.DataFrame(temuan_missing)],
        ignore_index=True
    )

# Hilangkan temuan identik jika sel dijalankan ulang
profil_tim = profil_tim.drop_duplicates().reset_index(drop=True)

# Urutkan berdasarkan tingkat keparahan
urutan = {'Tinggi': 0, 'Sedang': 1, 'Rendah': 2}

profil_tim['_urutan'] = (
    profil_tim['keparahan'].map(urutan).fillna(3)
)

profil_tim = (
    profil_tim.sort_values(
        ['_urutan', 'jumlah'],
        ascending=[True, False]
    )
    .drop(columns='_urutan')
    .reset_index(drop=True)
)

# Simpan laporan
os.makedirs('reports', exist_ok=True)

profil_tim.to_csv(
    'reports/profil_kualitas_data.csv',
    index=False
)

print("PROFIL KUALITAS DATA PROYEK TIM")
print("=" * 60)
print("Jumlah baris dataset:", len(df_tim))
print("Jumlah kolom dataset:", len(df_tim.columns))
print("Jumlah temuan dalam laporan:", len(profil_tim))
print("\nRingkasan missing value:")
display(
    pd.DataFrame({
        'jumlah_kosong': df_tim.isna().sum(),
        'persen_kosong': (df_tim.isna().mean() * 100).round(2)
    }).query('persen_kosong > 0').sort_values(
        'persen_kosong', ascending=False
    )
)

print("\nTabel profil kualitas data:")
display(profil_tim)

print("\nLaporan tersimpan di reports/profil_kualitas_data.csv")
````
